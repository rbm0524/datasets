```python
from transformers import Trainer, TrainingArguments
from peft import LoraConfig, get_peft_model, TaskType
from datasets import load_dataset
import torch

# 1. 모델에 LoRA 적용
lora_config = LoraConfig(
    r=8,
    lora_alpha=32,
    target_modules=["q_proj", "v_proj"],  # llama-3 구조에 맞는 모듈
    lora_dropout=0.05,
    bias="none",
    task_type=TaskType.CAUSAL_LM,
)

model = get_peft_model(model, lora_config)
model.print_trainable_parameters()

# 2. 데이터셋 Hugging Face Datasets로 불러오기
dataset = load_dataset("json", data_files="/content/drive/My Drive/final_capstone/dataset/llama_dialog_dataset_final.jsonl")["train"]

# 3. train : test 8:2 분할
split_dataset = dataset.train_test_split(test_size=0.2, seed=42)
train_dataset = split_dataset["train"]
eval_dataset = split_dataset["test"]

# 4. 전처리 함수
def preprocess_function(examples):
    prompts = [
        instruction + "\n" + inp if inp else instruction
        for instruction, inp in zip(examples["instruction"], examples["input"])
    ]

    model_inputs = tokenizer(prompts, truncation=True, padding="max_length", max_length=512)

    with tokenizer.as_target_tokenizer():
        labels = tokenizer(examples["output"], truncation=True, padding="max_length", max_length=512)

    model_inputs["labels"] = labels["input_ids"]
    return model_inputs

# 5. 전처리 적용
tokenized_train = train_dataset.map(preprocess_function, batched=True)
tokenized_eval = eval_dataset.map(preprocess_function, batched=True)

# 6. 학습 인자 설정
training_args = TrainingArguments(
    output_dir="./lora-llama-3-korean",
    per_device_train_batch_size=1,
    gradient_accumulation_steps=4,
    learning_rate=1e-4,
    num_train_epochs=3,
    fp16=torch.cuda.is_available(),
    logging_steps=100,
    save_steps=1000,
    evaluation_strategy="steps",
    eval_steps=500,
    load_best_model_at_end=True,
    report_to="none"
)

# 7. Trainer 구성 및 학습 시작
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train,
    eval_dataset=tokenized_eval
)

trainer.train()

# 8. 학습된 모델 저장
trainer.save_model("./lora-llama-3-korean")
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer, Trainer, TrainingArguments
from peft import PeftModel, PeftConfig
import torch

# 1. PEFT 구성 및 base 모델 로드
peft_model_path = "./lora-llama-3-korean"
config = PeftConfig.from_pretrained(peft_model_path)

base_model = AutoModelForCausalLM.from_pretrained(config.base_model_name_or_path, torch_dtype=torch.float16)
model = PeftModel.from_pretrained(base_model, peft_model_path)
model.eval()

# 2. 토크나이저 로드
tokenizer = AutoTokenizer.from_pretrained(config.base_model_name_or_path)

# 3. 전처리 함수 (기존과 동일)
def preprocess_function(examples):
    prompts = [
        instruction + "\n" + inp if inp else instruction
        for instruction, inp in zip(examples["instruction"], examples["input"])
    ]

    model_inputs = tokenizer(prompts, truncation=True, padding="max_length", max_length=512)

    with tokenizer.as_target_tokenizer():
        labels = tokenizer(examples["output"], truncation=True, padding="max_length", max_length=512)

    model_inputs["labels"] = labels["input_ids"]
    return model_inputs

# 4. 기존 평가 데이터셋에 전처리 적용
tokenized_eval = eval_dataset.map(preprocess_function, batched=True)

# 5. 평가용 Trainer 설정
eval_args = TrainingArguments(
    output_dir="./eval_results",
    per_device_eval_batch_size=1,
    report_to="none",
)

trainer = Trainer(
    model=model,
    args=eval_args,
    eval_dataset=tokenized_eval
)

# 6. 평가 수행
eval_results = trainer.evaluate()
print("Evaluation Results:", eval_results)
```
